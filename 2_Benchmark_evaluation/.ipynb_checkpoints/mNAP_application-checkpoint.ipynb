# MetFrag and mNAP application example

This notebook is the application workflow corresponding to the NIST benchmark. Run all cells in order from top to bottom. It directly loads the local GNPS files, generates MetFrag candidates for unknown nodes, applies mNAP, exports the rankings, and draws the top candidate structures.

The fixed calculations are identical to those used in the benchmark: 10 ppm MetFrag mass tolerance, up to 50 unique 2D candidates per node, MFP2-bits structural fingerprints, sigmoid parameters `a = -9` and `b = 0.60`, `beta = 0.5`, score floor `0.05`, and random-walk restart probability `0.5`.


```python
# =========================================================
# 1. Imports
# =========================================================
from __future__ import annotations

from pathlib import Path
import json
from typing import Dict, List, Optional, Sequence, Tuple, Union

import networkx as nx
import numpy as np
import pandas as pd
from IPython.display import display
from pyteomics import mgf
from rdkit import Chem, DataStructs, RDLogger
from rdkit.Chem import Draw, rdMolDescriptors
from matchms import Spectrum

try:
    from chemwalker.rwalker import random_walk
except Exception as exc:
    raise ImportError("Cannot import random_walk from chemwalker.rwalker.") from exc

try:
    from chemwalker.utils import run_metfrag
except Exception:
    try:
        from chemwalker.rwalker import run_metfrag
    except Exception as exc:
        raise ImportError("Cannot import run_metfrag from the installed ChemWalker package.") from exc

RDLogger.DisableLog("rdApp.warning")
```

```python
# =========================================================
# 2. Paths and fixed parameters
# =========================================================
# Edit only this section before running the notebook.
PROJECT_DIR = Path(r"D:\suwenyuan\Benchmark_evaluation\mnap_application")

CLUSTERINFO_PATH = PROJECT_DIR / "clusterinfo_summary.tsv"   # optional, used only for inspection
NETWORK_PATH = PROJECT_DIR / "networking_pairs.tsv"
SPECTRA_PATH = PROJECT_DIR / "spectra.mgf"
SEED_PATH = PROJECT_DIR / "seed_nodes.csv"
PUBCHEMLITE_PATH = Path(r"C:\Users\ZJM\Desktop\PubChemLite.csv")
METFRAG_PATH = PROJECT_DIR / "MetFrag2.3-CL.jar"              # use the same JAR as in the benchmark
OUTPUT_DIR = PROJECT_DIR / "results"

# Select one component by ComponentIndex, or provide an explicit node list.
COMPONENT_ID: Optional[int] = 6164
COMPONENT_NODES: Optional[List[int]] = None                    # example: [88874, 95957]
TARGET_NODE = 88874

# Rebuild the selected component from the MGF spectra so that cosine and matched
# product-ion thresholds are verified exactly as in the benchmark.
REBUILD_COMPONENT_NETWORK = True

# Molecular-network parameters
FRAGMENT_TOLERANCE = 0.01
MIN_COSINE = 0.60
MIN_MATCHED_PEAKS = 2
MUTUAL_TOPK = 10
MZ_POWER = 0.0
INTENSITY_POWER = 1.0
USE_HUNGARIAN = False

# MetFrag parameters
PPM = 10.0
TOP_N_CANDIDATES = 50
ADDUCT = "[M+H]+"
DEDUPLICATE_BY_INCHIKEY1 = True

# mNAP parameters
FINGERPRINT_METHOD = "MFP2-bits"
SIGMOID_A = -9.0
SIGMOID_B = 0.60
MNAP_BETA = 0.5
MNAP_SCORE_EPS = 0.05
RESTART_PROBABILITY = 0.5

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
METFRAG_AUDIT_DIR = OUTPUT_DIR / "metfrag_score_audit"
METFRAG_AUDIT_DIR.mkdir(parents=True, exist_ok=True)
```

```python
# =========================================================
# 3. Load local GNPS files, spectra, seed annotations and PubChemLite
# =========================================================
required_paths = [NETWORK_PATH, SPECTRA_PATH, SEED_PATH, PUBCHEMLITE_PATH, METFRAG_PATH]
missing_paths = [path for path in required_paths if not Path(path).exists()]
if missing_paths:
    raise FileNotFoundError(f"Missing required files: {missing_paths}")

net = pd.read_csv(NETWORK_PATH, sep="\t")
net.columns = [str(column).strip() for column in net.columns]

spectra = []
with mgf.MGF(str(SPECTRA_PATH)) as reader:
    spectra.extend(reader)
if not spectra:
    raise ValueError(f"No spectra were read from {SPECTRA_PATH}")

match_tab = pd.read_csv(SEED_PATH)
match_tab.columns = [str(column).strip() for column in match_tab.columns]

raw_db = pd.read_csv(PUBCHEMLITE_PATH)
raw_db.columns = [str(column).strip() for column in raw_db.columns]

gnps_tab = (
    pd.read_csv(CLUSTERINFO_PATH, sep="\t")
    if CLUSTERINFO_PATH.exists()
    else pd.DataFrame()
)

print(f"Loaded {len(net)} GNPS network edges, {len(spectra)} spectra, "
      f"{len(match_tab)} seed rows, and {len(raw_db)} PubChemLite rows.")
print("Network columns:", net.columns.tolist())
print("Seed columns:", match_tab.columns.tolist())
if not gnps_tab.empty:
    print(f"Cluster information contains {len(gnps_tab)} nodes.")

display(net.head())
display(match_tab.head())
```

```python
# =========================================================
# 4. General utilities and molecular-network functions
# =========================================================
def clear_df_attrs(df: pd.DataFrame) -> pd.DataFrame:
    out = df.copy()
    out.attrs.clear()
    return out


def clean_str(value) -> str:
    if value is None:
        return ""
    try:
        if pd.isna(value):
            return ""
    except Exception:
        pass
    return str(value).strip().strip('"').strip()


def normalize_node_id(value) -> int:
    text = clean_str(value)
    if text == "":
        raise ValueError("Missing node identifier.")
    try:
        return int(float(text))
    except Exception as exc:
        raise ValueError(f"Invalid node identifier: {value}") from exc


def normalize_inchi(value) -> str:
    text = clean_str(value)
    if text == "" or text.lower() in {"nan", "none", "null", "n/a", "na"}:
        return ""
    position = text.find("InChI=")
    if position >= 0:
        text = text[position:].split()[0]
    return text if text.startswith("InChI=") else "InChI=" + text


def smiles_to_inchi(smiles) -> str:
    try:
        text = clean_str(smiles)
        if text == "":
            return ""
        molecule = Chem.MolFromSmiles(text)
        return "" if molecule is None else Chem.MolToInchi(molecule)
    except Exception:
        return ""


def inchi_to_inchikey(inchi) -> str:
    try:
        text = normalize_inchi(inchi)
        if not text.startswith("InChI="):
            return ""
        value = Chem.InchiToInchiKey(text)
        return "" if value is None else value
    except Exception:
        return ""


def inchikey1(value) -> str:
    text = clean_str(value)
    return "" if text == "" else text.split("-")[0]


def mol_from_inchi(inchi):
    try:
        text = normalize_inchi(inchi)
        return Chem.MolFromInchi(text) if text.startswith("InChI=") else None
    except Exception:
        return None


def to_float(value, default=np.nan):
    try:
        if value is None:
            return default
        if isinstance(value, (list, tuple, np.ndarray)):
            if len(value) == 0:
                return default
            value = value[0]
        return float(str(value).split()[0])
    except Exception:
        return default


def pick_col(df: pd.DataFrame, candidates: Sequence[str]):
    lower_map = {str(column).lower(): column for column in df.columns}
    for candidate in candidates:
        if candidate.lower() in lower_map:
            return lower_map[candidate.lower()]
    for column in df.columns:
        lower = str(column).lower()
        if any(candidate.lower() in lower for candidate in candidates):
            return column
    return None


def spectrum_node_id(spectrum: dict) -> int:
    params = spectrum.get("params", {})
    for key in ["scans", "feature_id", "featureid"]:
        if key in params and clean_str(params[key]) != "":
            return normalize_node_id(params[key])
    raise ValueError("Spectrum metadata does not contain scans or feature_id.")


def spectrum_precursor_mz(spectrum: dict) -> float:
    precursor = to_float(spectrum.get("params", {}).get("pepmass"))
    if pd.isna(precursor):
        raise ValueError(f"Spectrum {spectrum_node_id(spectrum)} lacks a precursor m/z.")
    return float(precursor)


def index_spectra_by_node(spectra_list: Sequence[dict]) -> Dict[int, dict]:
    index: Dict[int, dict] = {}
    for spectrum in spectra_list:
        node_id = spectrum_node_id(spectrum)
        if node_id in index:
            raise ValueError(f"Multiple spectra share node identifier {node_id}.")
        index[node_id] = spectrum
    return index


def select_component_nodes(
    network: pd.DataFrame,
    component_id: Optional[int],
    explicit_nodes: Optional[Sequence[int]],
) -> Tuple[List[int], pd.DataFrame]:
    network = network.copy()
    required = ["CLUSTERID1", "CLUSTERID2", "Cosine"]
    missing = [column for column in required if column not in network.columns]
    if missing:
        raise ValueError(f"Network table is missing required columns: {missing}")
    network["CLUSTERID1"] = network["CLUSTERID1"].map(normalize_node_id)
    network["CLUSTERID2"] = network["CLUSTERID2"].map(normalize_node_id)
    network["Cosine"] = pd.to_numeric(network["Cosine"], errors="coerce")
    network = network.dropna(subset=["Cosine"]).copy()

    if explicit_nodes is not None:
        nodes = sorted({normalize_node_id(value) for value in explicit_nodes})
        subset = network[
            network["CLUSTERID1"].isin(nodes) & network["CLUSTERID2"].isin(nodes)
        ].copy()
    else:
        if component_id is None:
            raise ValueError("Set COMPONENT_ID or COMPONENT_NODES.")
        if "ComponentIndex" not in network.columns:
            raise ValueError("ComponentIndex is required when COMPONENT_ID is used.")
        component_values = pd.to_numeric(network["ComponentIndex"], errors="coerce")
        subset = network[component_values.eq(int(component_id))].copy()
        nodes = sorted(
            set(subset["CLUSTERID1"].astype(int)).union(
                set(subset["CLUSTERID2"].astype(int))
            )
        )
    if subset.empty or len(nodes) < 2:
        raise ValueError("The selected component contains fewer than two connected nodes.")
    return nodes, subset.reset_index(drop=True)


def get_modified_cosine_scorer():
    import matchms
    attempts = []
    if USE_HUNGARIAN:
        import_paths = [
            ("matchms.similarity", "ModifiedCosineHungarian"),
            ("matchms.similarity.ModifiedCosineHungarian", "ModifiedCosineHungarian"),
        ]
    else:
        import_paths = [
            ("matchms.similarity", "ModifiedCosineGreedy"),
            ("matchms.similarity.ModifiedCosineGreedy", "ModifiedCosineGreedy"),
            ("matchms.similarity", "ModifiedCosine"),
            ("matchms.similarity.ModifiedCosine", "ModifiedCosine"),
        ]
    for module_name, class_name in import_paths:
        try:
            module = __import__(module_name, fromlist=[class_name])
            scorer_class = getattr(module, class_name)
            scorer = scorer_class(
                tolerance=FRAGMENT_TOLERANCE,
                mz_power=MZ_POWER,
                intensity_power=INTENSITY_POWER,
            )
            return scorer, f"{class_name} (matchms {getattr(matchms, '__version__', 'unknown')})"
        except Exception as exc:
            attempts.append(f"{module_name}.{class_name}: {type(exc).__name__}: {exc}")
    raise ImportError("No compatible modified-cosine scorer was found:\n- " + "\n- ".join(attempts))


def get_score_and_matches(score_result) -> Tuple[float, int]:
    try:
        return float(score_result["score"]), int(score_result["matches"])
    except Exception:
        pass
    try:
        return float(score_result[0]), int(score_result[1])
    except Exception as exc:
        raise ValueError(f"Cannot parse matchms score result: {score_result}") from exc


def pyteomics_to_matchms(spectrum: dict) -> Spectrum:
    return Spectrum(
        mz=np.asarray(spectrum.get("m/z array", []), dtype=float),
        intensities=np.asarray(spectrum.get("intensity array", []), dtype=float),
        metadata={
            "precursor_mz": spectrum_precursor_mz(spectrum),
            "node_id": spectrum_node_id(spectrum),
        },
    )


def rebuild_component_network(
    component_nodes: Sequence[int], spectra_by_node: Dict[int, dict]
) -> pd.DataFrame:
    missing = sorted(set(component_nodes) - set(spectra_by_node))
    if missing:
        raise ValueError(f"Missing spectra for component nodes: {missing}")
    scorer, scorer_name = get_modified_cosine_scorer()
    print("Rebuilding the selected component with", scorer_name)
    selected = [
        (
            int(node),
            pyteomics_to_matchms(spectra_by_node[int(node)]),
            spectrum_precursor_mz(spectra_by_node[int(node)]),
        )
        for node in component_nodes
    ]
    rows = []
    for i in range(len(selected)):
        for j in range(i + 1, len(selected)):
            node1, spectrum1, precursor1 = selected[i]
            node2, spectrum2, precursor2 = selected[j]
            result = scorer.pair(spectrum1, spectrum2)
            cosine, matches = get_score_and_matches(result)
            if cosine >= MIN_COSINE and matches >= MIN_MATCHED_PEAKS:
                rows.append({
                    "CLUSTERID1": node1,
                    "CLUSTERID2": node2,
                    "Cosine": float(cosine),
                    "matched_peaks": int(matches),
                    "DeltaMZ": float(precursor2 - precursor1),
                })
    return pd.DataFrame(rows)


def filter_exported_network(network: pd.DataFrame) -> pd.DataFrame:
    out = network.copy()
    out["Cosine"] = pd.to_numeric(out["Cosine"], errors="coerce")
    out = out[out["Cosine"].ge(MIN_COSINE)].copy()
    matched_col = pick_col(
        out, ["matched_peaks", "MatchedPeaks", "SharedPeaks", "num_matched_peaks"]
    )
    if matched_col is not None:
        matched = pd.to_numeric(out[matched_col], errors="coerce")
        out = out[matched.ge(MIN_MATCHED_PEAKS)].copy()
    else:
        print(
            "The exported GNPS edge table has no matched-peak column. "
            "Set REBUILD_COMPONENT_NETWORK=True to verify the benchmark threshold."
        )
    return out.reset_index(drop=True)


def apply_mutual_topk(edges: pd.DataFrame, topk: int) -> pd.DataFrame:
    if edges.empty:
        return edges.copy()
    neighbors: Dict[int, List[Tuple[int, float]]] = {}
    for row in edges.itertuples(index=False):
        source, target, score = int(row.CLUSTERID1), int(row.CLUSTERID2), float(row.Cosine)
        neighbors.setdefault(source, []).append((target, score))
        neighbors.setdefault(target, []).append((source, score))
    top_neighbors = {
        node: {target for target, _ in sorted(values, key=lambda x: (-x[1], x[0]))[:topk]}
        for node, values in neighbors.items()
    }
    keep = []
    for row in edges.itertuples(index=False):
        source, target = int(row.CLUSTERID1), int(row.CLUSTERID2)
        keep.append(
            target in top_neighbors.get(source, set())
            and source in top_neighbors.get(target, set())
        )
    return edges.loc[keep].copy().reset_index(drop=True)
```

```python
# =========================================================
# 5. Select and prepare the component network
# =========================================================
spectra_by_node = index_spectra_by_node(spectra)
component_nodes, gnps_component_network = select_component_nodes(
    net, component_id=COMPONENT_ID, explicit_nodes=COMPONENT_NODES
)

if REBUILD_COMPONENT_NETWORK:
    component_network = rebuild_component_network(component_nodes, spectra_by_node)
else:
    component_network = filter_exported_network(gnps_component_network)

component_network = apply_mutual_topk(component_network, MUTUAL_TOPK)
if component_network.empty:
    raise ValueError(
        "No edge remained after applying cosine, matched-peak and mutual top-k filters."
    )

connected_nodes = set(component_network["CLUSTERID1"].astype(int)).union(
    set(component_network["CLUSTERID2"].astype(int))
)
disconnected_nodes = sorted(set(component_nodes) - connected_nodes)
if disconnected_nodes:
    raise ValueError(
        "Selected nodes became disconnected under the benchmark network criteria: "
        f"{disconnected_nodes}"
    )

print(f"Selected component nodes ({len(component_nodes)}): {component_nodes}")
print(f"Retained component edges: {len(component_network)}")
display(component_network)
```

```python
# =========================================================
# 6. Standardize PubChemLite and prepare verified seed structures
# =========================================================
def standardize_pubchemlite_db(source: Union[pd.DataFrame, str, Path]) -> pd.DataFrame:
    raw = source.copy() if isinstance(source, pd.DataFrame) else pd.read_csv(source)
    raw.columns = [str(column).strip() for column in raw.columns]
    id_col = pick_col(raw, ["Identifier", "CID", "PubChemCID", "compound_id", "id"])
    mass_col = pick_col(
        raw, ["MonoisotopicMass", "monoisotopic_mass", "exactmass", "ExactMass", "mass"]
    )
    inchi_col = pick_col(raw, ["InChI", "inchi", "INCHI"])
    smiles_col = pick_col(raw, ["SMILES", "smiles", "CanonicalSMILES", "canonical_smiles"])
    formula_col = pick_col(raw, ["MolecularFormula", "formula", "molecular_formula"])
    if id_col is None or mass_col is None:
        raise ValueError("PubChemLite requires Identifier and MonoisotopicMass columns.")
    if inchi_col is None and smiles_col is None:
        raise ValueError("PubChemLite requires InChI or SMILES.")

    db = pd.DataFrame()
    db["Identifier"] = raw[id_col].astype(str)
    db["MonoisotopicMass"] = pd.to_numeric(raw[mass_col], errors="coerce")
    db["MolecularFormula"] = raw[formula_col].astype(str) if formula_col is not None else ""
    db["SMILES"] = raw[smiles_col].astype(str) if smiles_col is not None else ""
    db["InChI"] = raw[inchi_col].apply(normalize_inchi) if inchi_col is not None else ""
    bad_inchi = ~db["InChI"].map(lambda value: mol_from_inchi(value) is not None)
    if bad_inchi.any() and smiles_col is not None:
        db.loc[bad_inchi, "InChI"] = db.loc[bad_inchi, "SMILES"].apply(smiles_to_inchi)

    db = db.dropna(subset=["MonoisotopicMass"]).copy()
    db = db[db["InChI"].map(lambda value: mol_from_inchi(value) is not None)].copy()
    db["InChIKey"] = db["InChI"].apply(inchi_to_inchikey)
    db["InChIKey1"] = db["InChIKey"].apply(inchikey1)
    db = db[db["InChIKey1"].astype(str).str.len().eq(14)].copy()
    keep = [
        "Identifier", "MonoisotopicMass", "MolecularFormula", "SMILES",
        "InChI", "InChIKey", "InChIKey1",
    ]
    return db[keep].drop_duplicates("Identifier").reset_index(drop=True)


def prepare_seed_table(seed_source: pd.DataFrame, nodes: Sequence[int]) -> pd.DataFrame:
    raw = seed_source.copy()
    raw.columns = [str(column).strip() for column in raw.columns]
    node_col = pick_col(raw, ["cluster index", "cluster_index", "#Scan#", "scan", "node_id"])
    identifier_col = pick_col(
        raw, ["Identifier", "Compound_Name", "Compound name", "LibraryID", "name"]
    )
    inchi_col = pick_col(raw, ["InChI", "INCHI", "inchi"])
    smiles_col = pick_col(raw, ["SMILES", "smiles", "CanonicalSMILES", "IsomericSMILES"])
    if node_col is None:
        raise ValueError("Seed table requires cluster index, #Scan#, or node_id.")
    if inchi_col is None and smiles_col is None:
        raise ValueError("Seed table requires InChI or SMILES.")

    seed = pd.DataFrame()
    seed["node_id"] = raw[node_col].map(normalize_node_id)
    seed["SourceIdentifier"] = (
        raw[identifier_col].astype(str)
        if identifier_col is not None
        else seed["node_id"].astype(str)
    )
    seed["InChI"] = raw[inchi_col].apply(normalize_inchi) if inchi_col is not None else ""
    bad_inchi = ~seed["InChI"].map(lambda value: mol_from_inchi(value) is not None)
    if bad_inchi.any() and smiles_col is not None:
        seed.loc[bad_inchi, "InChI"] = raw.loc[bad_inchi, smiles_col].apply(smiles_to_inchi)
    bad_inchi = ~seed["InChI"].map(lambda value: mol_from_inchi(value) is not None)
    if bad_inchi.any():
        raise ValueError(
            "Seed rows without a valid InChI or SMILES:\n"
            + seed.loc[bad_inchi, ["node_id", "SourceIdentifier", "InChI"]].to_string(index=False)
        )

    seed = seed[seed["node_id"].isin({int(value) for value in nodes})].copy()
    if seed.empty:
        raise ValueError("No verified seed annotation belongs to the selected component.")
    if seed.duplicated("node_id").any():
        duplicated = seed.loc[seed.duplicated("node_id", keep=False), "node_id"].tolist()
        raise ValueError(f"Each seed node must have one verified structure; duplicates: {duplicated}")

    seed["InChIKey"] = seed["InChI"].apply(inchi_to_inchikey)
    seed["InChIKey1"] = seed["InChIKey"].apply(inchikey1)
    seed["Identifier"] = "LibraryMatch_" + seed["node_id"].astype(str)
    seed["Score"] = 1.0
    seed["score_norm"] = 1.0
    seed["is_seed"] = True
    seed["seed_type"] = "verified_spectral_library_match"
    seed["uid"] = seed["node_id"].astype(str) + "_" + seed["Identifier"]
    keep = [
        "uid", "node_id", "Identifier", "SourceIdentifier", "InChI", "InChIKey",
        "InChIKey1", "Score", "score_norm", "is_seed", "seed_type",
    ]
    return clear_df_attrs(seed[keep].reset_index(drop=True))


db = standardize_pubchemlite_db(raw_db)
seed_rows = prepare_seed_table(match_tab, component_nodes)
seed_nodes = set(seed_rows["node_id"].astype(int))
target_nodes = sorted(set(component_nodes) - seed_nodes)
if not target_nodes:
    raise ValueError("At least one non-seed target node is required.")

print(f"Standardized PubChemLite rows: {len(db)}")
print(f"Seed nodes: {sorted(seed_nodes)}")
print(f"Target nodes: {target_nodes}")
display(seed_rows)
```

```python
# =========================================================
# 7. Strict MetFrag output validation
# =========================================================
class MetFragOutputValidationError(RuntimeError):
    """Raised when the true MetFrag Score cannot be identified unambiguously."""


OFFICIAL_METFRAG_SCORE_COLUMN = "Score"
METFRAG_WRAPPER_INDEX_COLUMN = "__metfrag_wrapper_index__"
METFRAG_WRAPPER_INDEX_DEFAULT_COLUMN = "__metfrag_wrapper_index_is_default__"
METFRAG_SCORE_SOURCE_COLUMN = "Score column"
METFRAG_SCORE_SOURCE_INDEX = "DataFrame index (recovered PSV Score)"


def _make_unique_column_names(columns: Sequence[str]) -> List[str]:
    counts: Dict[str, int] = {}
    names: List[str] = []
    for column in columns:
        base = str(column).strip()
        counts[base] = counts.get(base, 0) + 1
        names.append(base if counts[base] == 1 else f"{base}__{counts[base]}")
    return names


def _base_name(column: str) -> str:
    return str(column).split("__", 1)[0]


def _columns_with_base(df: pd.DataFrame, base: str) -> List[str]:
    return [column for column in df.columns if _base_name(column) == base]


def _looks_like_normalized_metfrag_score(values: pd.Series) -> bool:
    numeric = pd.to_numeric(values, errors="coerce").dropna()
    return bool(
        not numeric.empty
        and numeric.min() >= -1e-12
        and numeric.max() <= 1.000001
        and np.isfinite(numeric).all()
    )


def normalize_metfrag_output(result: pd.DataFrame) -> pd.DataFrame:
    if result is None or len(result) == 0:
        return pd.DataFrame()
    df = clear_df_attrs(result)
    raw_index = df.index.copy()
    is_default_range = (
        isinstance(raw_index, pd.RangeIndex)
        and raw_index.start == 0
        and raw_index.stop == len(df)
        and raw_index.step == 1
    )
    df.columns = _make_unique_column_names(df.columns)
    df.insert(0, METFRAG_WRAPPER_INDEX_COLUMN, list(raw_index))
    df.insert(1, METFRAG_WRAPPER_INDEX_DEFAULT_COLUMN, bool(is_default_range))
    return df.reset_index(drop=True)


def extract_verified_metfrag_score(df: pd.DataFrame, node_id: int) -> Tuple[pd.Series, str]:
    score_columns = _columns_with_base(df, OFFICIAL_METFRAG_SCORE_COLUMN)
    if not score_columns:
        raise MetFragOutputValidationError(
            f"MetFrag output for node {node_id} has no visible Score column."
        )
    visible_column = max(
        score_columns,
        key=lambda column: pd.to_numeric(df[column], errors="coerce").notna().sum(),
    )
    visible_score = pd.to_numeric(df[visible_column], errors="coerce")
    index_values = pd.to_numeric(df[METFRAG_WRAPPER_INDEX_COLUMN], errors="coerce")
    index_is_default = bool(
        df[METFRAG_WRAPPER_INDEX_DEFAULT_COLUMN].astype(bool).all()
    )
    if _looks_like_normalized_metfrag_score(visible_score):
        return visible_score, METFRAG_SCORE_SOURCE_COLUMN
    if not index_is_default and _looks_like_normalized_metfrag_score(index_values):
        return index_values, METFRAG_SCORE_SOURCE_INDEX
    raise MetFragOutputValidationError(
        f"Unable to identify the true MetFrag Score for node {node_id}. "
        "The workflow stopped rather than substituting a mass, row number, or another score."
    )


def _find_best_inchi_column(df: pd.DataFrame) -> Optional[str]:
    candidates = []
    for column in df.columns:
        count = int(df[column].astype(str).str.contains("InChI=", regex=False, na=False).sum())
        candidates.append((count, int(_base_name(column) == "InChI"), column))
    candidates.sort(reverse=True)
    return None if not candidates or candidates[0][0] <= 0 else candidates[0][2]


def _normalize_identifier(value) -> str:
    text = clean_str(value)
    if text.endswith(".0"):
        try:
            return str(int(float(text)))
        except Exception:
            pass
    return text


def map_identifiers_from_database(
    candidates: pd.DataFrame, candidate_db: pd.DataFrame
) -> pd.DataFrame:
    out = candidates.copy()
    mapping = candidate_db.copy()
    mapping["Identifier"] = mapping["Identifier"].map(_normalize_identifier)
    key_map = mapping.drop_duplicates("InChIKey").set_index("InChIKey")["Identifier"].to_dict()
    inchi_map = mapping.drop_duplicates("InChI").set_index("InChI")["Identifier"].to_dict()
    out["Identifier"] = out["InChIKey"].map(key_map).fillna("")
    missing = out["Identifier"].eq("")
    out.loc[missing, "Identifier"] = out.loc[missing, "InChI"].map(inchi_map).fillna("")
    missing = out["Identifier"].eq("")
    if missing.any():
        out.loc[missing, "Identifier"] = [
            f"cand_{index}_{key}"
            for index, key in zip(out.index[missing], out.loc[missing, "InChIKey1"])
        ]
    return out


def mass_filter_db(database: pd.DataFrame, neutral_mass: float, ppm: float) -> pd.DataFrame:
    tolerance = neutral_mass * ppm * 1e-6
    return database[
        database["MonoisotopicMass"].between(
            neutral_mass - tolerance, neutral_mass + tolerance, inclusive="both"
        )
    ].copy()


def infer_neutral_mass(precursor_mz: float) -> float:
    if ADDUCT != "[M+H]+":
        raise ValueError("This application notebook is configured for [M+H]+ ions.")
    return float(precursor_mz) - 1.007276466812


def build_metfrag_input(spectrum: dict, node_id: int) -> dict:
    return {
        "m/z array": np.asarray(spectrum.get("m/z array", []), dtype=float),
        "intensity array": np.asarray(spectrum.get("intensity array", []), dtype=float),
        "params": {
            "pepmass": [spectrum_precursor_mz(spectrum)],
            "charge": [1],
            "scans": str(node_id),
            "feature_id": str(node_id),
            "title": clean_str(spectrum.get("params", {}).get("title", node_id)),
        },
    }


def clean_metfrag_result(
    result: pd.DataFrame, node_id: int, candidate_db: pd.DataFrame
) -> pd.DataFrame:
    df = normalize_metfrag_output(result)
    if df.empty:
        return pd.DataFrame()
    verified_score, score_source = extract_verified_metfrag_score(df, node_id)
    df["MetFragScoreRaw"] = pd.to_numeric(verified_score, errors="coerce")
    df["Score"] = df["MetFragScoreRaw"]
    df["score_source"] = score_source

    inchi_column = _find_best_inchi_column(df)
    if inchi_column is None:
        raise MetFragOutputValidationError(
            f"MetFrag returned rows for node {node_id}, but no field contained an InChI."
        )
    df["InChI"] = df[inchi_column].apply(normalize_inchi)
    df = df[df["InChI"].map(lambda value: mol_from_inchi(value) is not None)].copy()
    if df.empty:
        raise MetFragOutputValidationError(f"No valid candidate structure for node {node_id}.")

    df["node_id"] = int(node_id)
    df["InChIKey"] = df["InChI"].apply(inchi_to_inchikey)
    df["InChIKey1"] = df["InChIKey"].apply(inchikey1)
    df = df[df["InChIKey1"].astype(str).str.len().eq(14)].copy()
    df = map_identifiers_from_database(df, candidate_db)
    df = df.sort_values(["Score", "Identifier"], ascending=[False, True])
    if DEDUPLICATE_BY_INCHIKEY1:
        df = df.drop_duplicates("InChIKey1", keep="first")
    df = df.head(TOP_N_CANDIDATES).copy()

    df["Identifier"] = df["Identifier"].astype(str)
    df["uid"] = df["node_id"].astype(str) + "_" + df["Identifier"]
    df["score_norm"] = df["Score"] / df["Score"].max() if df["Score"].max() > 0 else 0.0
    df["is_seed"] = False
    df["seed_type"] = ""
    keep = [
        "uid", "node_id", "Identifier", "InChI", "InChIKey", "InChIKey1",
        "Score", "MetFragScoreRaw", "score_source", "score_norm",
        "is_seed", "seed_type",
    ]
    return clear_df_attrs(df[keep].reset_index(drop=True))


def generate_metfrag_candidates(
    spectrum: dict, node_id: int, database: pd.DataFrame
) -> Tuple[pd.DataFrame, Dict[str, object]]:
    precursor_mz = spectrum_precursor_mz(spectrum)
    neutral_mass = infer_neutral_mass(precursor_mz)
    candidate_db = mass_filter_db(database, neutral_mass, PPM)
    status = {
        "node_id": int(node_id),
        "precursor_mz": precursor_mz,
        "neutral_mass": neutral_mass,
        "mass_filtered_database_rows": int(len(candidate_db)),
        "n_candidates": 0,
        "status": "no_mass_candidates" if candidate_db.empty else "pending",
    }
    if candidate_db.empty:
        return pd.DataFrame(), status

    raw_result = run_metfrag(
        build_metfrag_input(spectrum, node_id),
        candidate_db,
        int(node_id),
        metpath=str(METFRAG_PATH),
        ppm=PPM,
    )
    if raw_result is None or len(raw_result) == 0:
        status["status"] = "no_metfrag_candidates"
        return pd.DataFrame(), status

    cleaned = clean_metfrag_result(raw_result, node_id, candidate_db)
    normalize_metfrag_output(raw_result).to_csv(
        METFRAG_AUDIT_DIR / f"node_{node_id}_raw.csv", index=False
    )
    cleaned.to_csv(METFRAG_AUDIT_DIR / f"node_{node_id}_cleaned.csv", index=False)
    status["n_candidates"] = int(len(cleaned))
    status["status"] = "ok" if not cleaned.empty else "no_valid_candidates"
    return cleaned, status
```

```python
# =========================================================
# 8. Generate MetFrag candidates for all unknown nodes
# =========================================================
candidate_frames: List[pd.DataFrame] = []
status_rows: List[Dict[str, object]] = []

for index, node_id in enumerate(target_nodes, start=1):
    print(f"MetFrag target {index}/{len(target_nodes)}: node {node_id}")
    if node_id not in spectra_by_node:
        raise ValueError(f"No MS/MS spectrum was found for target node {node_id}.")
    candidates, status = generate_metfrag_candidates(
        spectra_by_node[node_id], node_id, db
    )
    status_rows.append(status)
    if not candidates.empty:
        candidate_frames.append(candidates)

metfrag_status = pd.DataFrame(status_rows)
if not candidate_frames:
    raise ValueError("MetFrag returned no valid candidate list for any target node.")

metfrag_candidates = clear_df_attrs(
    pd.concat(candidate_frames, ignore_index=True, sort=False)
)
print(f"MetFrag generated {len(metfrag_candidates)} candidate rows for "
      f"{metfrag_candidates['node_id'].nunique()} target nodes.")
display(metfrag_status)
display(metfrag_candidates.head(20))
```

```python
# =========================================================
# 9. mNAP candidate graph, random walk and ranking
# =========================================================
_FP_CACHE: Dict[Tuple[str, str], object] = {}


def get_structural_fingerprint(inchi, method: str = FINGERPRINT_METHOD):
    text = normalize_inchi(inchi)
    cache_key = (text, method)
    if cache_key in _FP_CACHE:
        return _FP_CACHE[cache_key]
    molecule = mol_from_inchi(text)
    if molecule is None:
        _FP_CACHE[cache_key] = None
        return None
    if method != "MFP2-bits":
        raise ValueError("The published mNAP calculation uses MFP2-bits.")
    fingerprint = rdMolDescriptors.GetHashedMorganFingerprint(molecule, 2)
    _FP_CACHE[cache_key] = fingerprint
    return fingerprint


def tanimoto(inchi1, inchi2) -> float:
    fp1 = get_structural_fingerprint(inchi1)
    fp2 = get_structural_fingerprint(inchi2)
    if fp1 is None or fp2 is None:
        return np.nan
    return float(DataStructs.TanimotoSimilarity(fp1, fp2))


def mnap_sigmoid(value: float) -> float:
    return float(1.0 / (1.0 + np.exp(SIGMOID_A * (value - SIGMOID_B))))


def build_mnap_candidate_edges(
    spectral_edges: pd.DataFrame, full_candidates: pd.DataFrame
) -> pd.DataFrame:
    groups = {
        int(node_id): clear_df_attrs(group)
        for node_id, group in full_candidates.groupby("node_id")
    }
    rows = []
    for edge in spectral_edges.itertuples(index=False):
        node1, node2, cosine = int(edge.CLUSTERID1), int(edge.CLUSTERID2), float(edge.Cosine)
        candidates1, candidates2 = groups.get(node1), groups.get(node2)
        if candidates1 is None or candidates2 is None:
            continue
        for candidate1 in candidates1.itertuples(index=False):
            for candidate2 in candidates2.itertuples(index=False):
                structure_similarity = tanimoto(candidate1.InChI, candidate2.InChI)
                if pd.isna(structure_similarity):
                    continue
                score1 = max(float(candidate1.score_norm), MNAP_SCORE_EPS)
                score2 = max(float(candidate2.score_norm), MNAP_SCORE_EPS)
                similarity_term = mnap_sigmoid(cosine * structure_similarity)
                weight = (
                    similarity_term
                    * (score1 ** MNAP_BETA)
                    * (score2 ** MNAP_BETA)
                )
                if weight > 0 and np.isfinite(weight):
                    rows.append({
                        "source": str(candidate1.uid),
                        "target": str(candidate2.uid),
                        "weight": float(weight),
                        "node1": node1,
                        "node2": node2,
                        "cosine": cosine,
                        "structural_similarity": float(structure_similarity),
                        "similarity_term": float(similarity_term),
                        "score1_norm": float(candidate1.score_norm),
                        "score2_norm": float(candidate2.score_norm),
                        "score1_effective": score1,
                        "score2_effective": score2,
                    })
    return pd.DataFrame(rows)


def mNAP(
    spectral_edges: pd.DataFrame,
    full_candidates: pd.DataFrame,
    seed_uids: Sequence[str],
) -> Tuple[pd.DataFrame, pd.DataFrame, nx.Graph]:
    """Apply the published mNAP edge equation and random-walk ranking."""
    candidate_edges = build_mnap_candidate_edges(spectral_edges, full_candidates)
    if candidate_edges.empty:
        raise ValueError("No candidate-level mNAP edge was generated.")

    graph = nx.Graph()
    graph.add_weighted_edges_from(
        candidate_edges[["source", "target", "weight"]].itertuples(
            index=False, name=None
        )
    )
    valid_seeds = [str(seed) for seed in seed_uids if str(seed) in graph.nodes]
    if not valid_seeds:
        raise ValueError("No verified seed structure is connected to the mNAP graph.")

    probabilities = random_walk(
        graph, valid_seeds, restart_prob=RESTART_PROBABILITY
    )
    if isinstance(probabilities, dict):
        probability_map = {str(node): float(value) for node, value in probabilities.items()}
    else:
        probability_map = {
            str(node): float(value)
            for node, value in zip(graph.nodes(), np.asarray(probabilities, dtype=float))
        }

    ranked = clear_df_attrs(full_candidates)
    ranked["mNAP_score"] = ranked["uid"].astype(str).map(probability_map).fillna(0.0)
    group_max = ranked.groupby("node_id")["mNAP_score"].transform("max")
    ranked["mNAP_reachable"] = group_max > 0
    ranked["mNAP_score_norm"] = np.where(
        group_max > 0, ranked["mNAP_score"] / group_max, 0.0
    )
    ranked["mNAP_rank"] = ranked.groupby("node_id")["mNAP_score_norm"].rank(
        ascending=False, method="dense"
    )
    ranked.loc[~ranked["mNAP_reachable"], "mNAP_rank"] = np.nan
    ranked["MetFrag_score"] = ranked["Score"]
    ranked["MetFrag_rank"] = ranked.groupby("node_id")["MetFrag_score"].rank(
        ascending=False, method="dense"
    )
    ranked["cluster index"] = ranked["node_id"].astype(int)
    ranked = ranked.sort_values(
        ["node_id", "mNAP_rank", "MetFrag_rank", "Identifier"],
        na_position="last",
    ).reset_index(drop=True)
    return ranked, candidate_edges, graph


# Combine one verified structure per seed node with MetFrag candidates for targets.
full_tlid = clear_df_attrs(
    pd.concat([seed_rows, metfrag_candidates], ignore_index=True, sort=False)
)
full_tlid["Score"] = pd.to_numeric(full_tlid["Score"], errors="coerce").fillna(0.0)
full_tlid["score_norm"] = full_tlid.groupby("node_id")["Score"].transform(
    lambda values: values / values.max() if values.max() > 0 else 0.0
)
full_tlid.loc[full_tlid["is_seed"].fillna(False).astype(bool), "score_norm"] = 1.0
full_tlid["uid"] = (
    full_tlid["node_id"].astype(int).astype(str)
    + "_"
    + full_tlid["Identifier"].astype(str)
)

invalid_structure = ~full_tlid["InChI"].map(lambda value: mol_from_inchi(value) is not None)
if invalid_structure.any():
    raise ValueError(
        "Invalid structures remained in the candidate table:\n"
        + full_tlid.loc[invalid_structure, ["node_id", "Identifier", "InChI"]]
        .head(20).to_string(index=False)
    )

result_tab, candidate_edges, mnap_graph = mNAP(
    spectral_edges=component_network,
    full_candidates=full_tlid,
    seed_uids=seed_rows["uid"].astype(str).tolist(),
)

print(f"mNAP candidate graph: {mnap_graph.number_of_nodes()} nodes and "
      f"{mnap_graph.number_of_edges()} edges.")
display(result_tab.head(20))
```

```python
# =========================================================
# 10. Inspect and export MetFrag and mNAP rankings
# =========================================================
result_columns = [
    "node_id", "Identifier", "InChIKey1", "MetFrag_score", "score_norm",
    "MetFrag_rank", "mNAP_score", "mNAP_score_norm", "mNAP_rank",
    "mNAP_reachable", "InChI",
]

target_result = result_tab[
    result_tab["node_id"].astype(int).eq(int(TARGET_NODE))
    & ~result_tab["is_seed"].fillna(False).astype(bool)
].copy()
if target_result.empty:
    raise ValueError(
        f"TARGET_NODE={TARGET_NODE} has no MetFrag candidate list or is a seed node."
    )

print(f"Target node {TARGET_NODE}: mNAP ranking")
display(
    target_result[result_columns]
    .sort_values(["mNAP_rank", "MetFrag_rank", "Identifier"], na_position="last")
    .head(30)
)

print(f"Target node {TARGET_NODE}: original MetFrag ranking")
display(
    target_result[result_columns]
    .sort_values(["MetFrag_rank", "mNAP_rank", "Identifier"], na_position="last")
    .head(30)
)

result_tab.to_csv(OUTPUT_DIR / "mnap_rankings.csv", index=False)
metfrag_candidates.to_csv(OUTPUT_DIR / "metfrag_candidates.csv", index=False)
seed_rows.to_csv(OUTPUT_DIR / "mnap_seed_structures.csv", index=False)
candidate_edges.to_csv(OUTPUT_DIR / "mnap_candidate_edges.tsv", sep="\t", index=False)
metfrag_status.to_csv(OUTPUT_DIR / "metfrag_candidate_status.csv", index=False)
component_network.to_csv(OUTPUT_DIR / "component_network.tsv", sep="\t", index=False)
target_result.sort_values("mNAP_rank", na_position="last").to_csv(
    OUTPUT_DIR / f"target_{TARGET_NODE}_mnap_rankings.csv", index=False
)
target_result.sort_values("MetFrag_rank", na_position="last").to_csv(
    OUTPUT_DIR / f"target_{TARGET_NODE}_metfrag_rankings.csv", index=False
)

manifest = {
    "algorithm": "mNAP",
    "component_id": COMPONENT_ID,
    "component_nodes": component_nodes,
    "target_node": TARGET_NODE,
    "rebuild_component_network": REBUILD_COMPONENT_NETWORK,
    "network_parameters": {
        "fragment_tolerance": FRAGMENT_TOLERANCE,
        "minimum_cosine": MIN_COSINE,
        "minimum_matched_peaks": MIN_MATCHED_PEAKS,
        "mutual_topk": MUTUAL_TOPK,
    },
    "metfrag_parameters": {
        "ppm": PPM,
        "top_n_candidates": TOP_N_CANDIDATES,
        "adduct": ADDUCT,
        "deduplicate_by_inchikey1": DEDUPLICATE_BY_INCHIKEY1,
    },
    "mnap_parameters": {
        "fingerprint_method": FINGERPRINT_METHOD,
        "sigmoid_a": SIGMOID_A,
        "sigmoid_b": SIGMOID_B,
        "beta": MNAP_BETA,
        "score_epsilon": MNAP_SCORE_EPS,
        "restart_probability": RESTART_PROBABILITY,
    },
}
(OUTPUT_DIR / "run_manifest.json").write_text(
    json.dumps(manifest, indent=2, ensure_ascii=False), encoding="utf-8"
)
print("Outputs saved to:", OUTPUT_DIR.resolve())
```

```python
# =========================================================
# 11. Draw the top MetFrag and mNAP candidates
# =========================================================
def draw_candidate_panel(
    candidate_table: pd.DataFrame,
    rank_column: str,
    title: str,
    top_n: int = 5,
):
    top = candidate_table.sort_values(
        [rank_column, "Identifier"], na_position="last"
    ).head(top_n)
    molecules, legends = [], []
    for row in top.itertuples(index=False):
        molecule = mol_from_inchi(row.InChI)
        if molecule is None:
            continue
        molecules.append(molecule)
        legends.append(
            f"ID: {row.Identifier}\n"
            f"MetFrag rank: {int(row.MetFrag_rank)}\n"
            f"mNAP rank: {int(row.mNAP_rank) if pd.notna(row.mNAP_rank) else 'NA'}"
        )
    if not molecules:
        raise ValueError(f"No drawable candidate structure was found for {title}.")
    print(title)
    return Draw.MolsToGridImage(
        molecules,
        molsPerRow=min(5, top_n),
        subImgSize=(280, 250),
        legends=legends,
        useSVG=False,
    )


metfrag_image = draw_candidate_panel(
    target_result, "MetFrag_rank", f"Target {TARGET_NODE}: MetFrag top candidates", top_n=5
)
display(metfrag_image)
metfrag_image.save(OUTPUT_DIR / f"target_{TARGET_NODE}_MetFrag_top5.png")

mnap_image = draw_candidate_panel(
    target_result, "mNAP_rank", f"Target {TARGET_NODE}: mNAP top candidates", top_n=5
)
display(mnap_image)
mnap_image.save(OUTPUT_DIR / f"target_{TARGET_NODE}_mNAP_top5.png")
```

```python
import re
import numpy as np
import pandas as pd
import networkx as nx

from rdkit import Chem, DataStructs
from rdkit.Chem import AllChem, MACCSkeys, rdMolDescriptors

from chemwalker.rwalker import random_walk
from chemwalker.utils import run_metfrag


# =========================================================
# 基础工具函数
# =========================================================

def _show(x):
    try:
        display(x)
    except Exception:
        print(x)


def normalize_id(x):
    """把 67730842.0 这类 ID 统一成 67730842"""
    if pd.isna(x):
        return ""
    s = str(x).strip()
    if s.lower() in ["", "nan", "none", "null"]:
        return ""
    s = re.sub(r"\.0$", "", s)
    return s


def clean_inchi_string(x):
    """从任意字符串里提取 InChI=..."""
    if pd.isna(x):
        return ""
    s = str(x).strip()
    idx = s.find("InChI=")
    if idx < 0:
        return ""
    s = s[idx:].strip()
    # InChI 通常不含空格；防止后面带上别的字段
    s = s.split()[0]
    return s


def smiles_to_inchi(smiles):
    """SMILES 转 InChI"""
    if pd.isna(smiles) or str(smiles).strip() == "":
        return ""
    try:
        mol = Chem.MolFromSmiles(str(smiles).strip())
        if mol is None:
            return ""
        return Chem.MolToInchi(mol)
    except Exception:
        return ""


def mol_from_inchi(inchi):
    """InChI 转 RDKit mol"""
    inchi = clean_inchi_string(inchi)
    if not inchi.startswith("InChI="):
        return None
    try:
        mol = Chem.MolFromInchi(inchi)
        return mol
    except Exception:
        return None


def is_usable_inchi(x):
    """判断 InChI 是否真的能被 RDKit 读取"""
    return mol_from_inchi(x) is not None


def inchi_to_inchikey(inchi):
    """将有效 InChI 转换为 InChIKey。"""
    try:
        inchi = clean_inchi_string(inchi)
        return Chem.InchiToInchiKey(inchi) if inchi.startswith("InChI=") else ""
    except Exception:
        return ""


def inchikey1(inchikey):
    """返回 InChIKey 第一段，用于二维连接结构层面的候选物去重。"""
    value = str(inchikey).strip()
    return value.split("-")[0] if value else ""


def find_smiles_col(df):
    """根据列名找 SMILES 列"""
    possible = [
        "SMILES", "Smiles", "smiles",
        "CanonicalSMILES", "IsomericSMILES",
        "CanonicalSmiles", "IsomericSmiles"
    ]

    for c in possible:
        if c in df.columns:
            return c

    for c in df.columns:
        if "smiles" in str(c).lower():
            return c

    return None


def find_inchi_col_by_content(df):
    """根据内容找真正含有 InChI= 的列"""
    best_col = None
    best_n = 0

    for c in df.columns:
        s = df[c].astype(str).map(clean_inchi_string)
        n = s.str.startswith("InChI=").sum()
        if n > best_n:
            best_n = n
            best_col = c

    return best_col if best_n > 0 else None


def extract_inchi_rowwise(df):
    """逐行扫描所有单元格，找 InChI=...，比相信列名更安全"""
    inchis = []

    for _, row in df.iterrows():
        found = ""
        for val in row.values:
            tmp = clean_inchi_string(val)
            if tmp.startswith("InChI="):
                found = tmp
                break
        inchis.append(found)

    return pd.Series(inchis, index=df.index)


def series_looks_like_ids(s):
    vals = s.map(normalize_id)
    vals = vals[vals != ""]

    if len(vals) == 0:
        return False

    # 全是 0 不行
    if (vals == "0").all():
        return False

    return True


def choose_identifier_series(res):
    """尽量稳健地提取候选物 Identifier"""
    preferred = [
        "Identifier", "CID", "PubChemCID", "PubChem_CID",
        "CompoundID", "DatabaseID", "ID", "index"
    ]

    for c in preferred:
        if c in res.columns:
            s = res[c].map(normalize_id)
            if series_looks_like_ids(s):
                return s

    # 如果没有明确列名，找一个像 PubChem CID 的长整数列
    best_col = None
    best_n = 0

    for c in res.columns:
        vals = res[c].map(normalize_id)
        n = vals.str.match(r"^\d{4,}$", na=False).sum()
        if n > best_n:
            best_n = n
            best_col = c

    if best_col is not None and best_n > 0:
        return res[best_col].map(normalize_id)

    return pd.Series([f"cand_{i}" for i in range(len(res))], index=res.index)


def make_unique_ids(raw_ids):
    unique_ids = []
    seen = {}

    for val in raw_ids:
        val = normalize_id(val)

        if val == "":
            val = "cand"

        if val in seen:
            seen[val] += 1
            unique_ids.append(f"{val}_{seen[val]}")
        else:
            seen[val] = 0
            unique_ids.append(val)

    return unique_ids


def find_best_db_id_col(db_df, ids):
    """在 db_df 里找最能匹配候选物 Identifier 的 ID 列"""
    ids = set([normalize_id(x) for x in ids if normalize_id(x) != ""])
    if len(ids) == 0:
        return None

    best_col = None
    best_n = 0

    for c in db_df.columns:
        vals = db_df[c].map(normalize_id)
        n = vals.isin(ids).sum()

        if n > best_n:
            best_n = n
            best_col = c

    return best_col if best_n > 0 else None


def pick_score_series(res, manual_score_col=None):
    """
    提取 MetFrag score。
    重要：不再盲目相信名为 Score 的列，因为你这里的 Score 被错误抓成了 536 m/z/mass。
    """
    if manual_score_col is not None and manual_score_col in res.columns:
        s = pd.to_numeric(res[manual_score_col], errors="coerce").fillna(0)
        return s, manual_score_col

    # 1. 优先用列名里明确含 score 的列，但要求数值范围合理
    score_like_cols = []

    preferred = [
        "FinalScore", "MetFragScore", "FragmenterScore",
        "CombinedScore", "Score"
    ]

    for c in preferred:
        if c in res.columns:
            score_like_cols.append(c)

    for c in res.columns:
        if "score" in str(c).lower() and c not in score_like_cols:
            score_like_cols.append(c)

    for c in score_like_cols:
        s = pd.to_numeric(res[c], errors="coerce")
        if s.notna().sum() == 0:
            continue

        s_valid = s.dropna()

        # MetFrag score 通常是 0-1 或 0-10；大于 10 的通常不是 score
        if s_valid.min() >= 0 and s_valid.max() <= 10:
            return s.fillna(0), c

    # 2. 如果列名错乱，就从所有数值列里找 0-1 且有变化的列
    candidates = []

    exclude_keywords = [
        "id", "identifier", "cid", "index",
        "mz", "mass", "molecularweight",
        "inchi", "smiles", "charge"
    ]

    for c in res.columns:
        cname = str(c).lower()

        if any(k in cname for k in exclude_keywords):
            continue

        s = pd.to_numeric(res[c], errors="coerce")

        if s.notna().sum() == 0:
            continue

        s_valid = s.dropna()

        if s_valid.min() >= 0 and s_valid.max() <= 1.000001:
            if s_valid.nunique() > 1:
                candidates.append((c, s.fillna(0), s_valid.std()))

    if len(candidates) > 0:
        candidates = sorted(candidates, key=lambda x: x[2], reverse=True)
        c, s, _ = candidates[0]
        return s, c

    # 3. 实在找不到就给 0
    return pd.Series([0.0] * len(res), index=res.index), None


# =========================================================
# MetFrag 候选物清洗
# =========================================================

def clean_metfrag_result(res, node_id, db_df=None, params=None, verbose=True):
    res = res.reset_index(drop=False).copy()
    res.columns = [str(c).strip() for c in res.columns]

    print(f"\n节点 {node_id} MetFrag 原始输出列名：")
    print(res.columns.tolist())

    # 针对你当前 run_metfrag 的错位输出：
    # index = 真正 MetFrag score
    # NumberPeaksUsed = 真正 InChI
    # MaximumTreeDepth = 真正 Identifier / CID

    if "NumberPeaksUsed" in res.columns and res["NumberPeaksUsed"].astype(str).str.startswith("InChI=").any():
        print("⚠️ 检测到 MetFrag 输出列错位，使用固定映射修正。")

        res["InChI"] = res["NumberPeaksUsed"].astype(str)
        res["Identifier"] = res["MaximumTreeDepth"].astype(str).str.replace(r"\.0$", "", regex=True)

        res["Score_Clean"] = pd.to_numeric(res["index"], errors="coerce")
        res["Score"] = res["Score_Clean"]

    else:
        raise ValueError("没有检测到预期的 InChI 列，请重新检查 MetFrag 输出。")

    # 与基准评估一致：正式分析只接受 0-1 范围内的 MetFrag Score
    valid_scores = res["Score_Clean"].dropna()
    if valid_scores.empty or valid_scores.min() < 0 or valid_scores.max() > 1.000001:
        raise ValueError(
            f"节点 {node_id} 的 MetFrag Score 不在预期的 0-1 范围内，请检查原始输出。"
        )

    # 过滤无效 InChI
    res = res[res["InChI"].map(is_usable_inchi)].copy()

    # 与基准评估一致：按 InChIKey 第一段进行二维结构去重，并保留前 50 个候选物
    res["InChIKey"] = res["InChI"].apply(inchi_to_inchikey)
    res["InChIKey1"] = res["InChIKey"].apply(inchikey1)
    res = res[res["InChIKey1"].astype(str).str.len().eq(14)].copy()
    res = res.sort_values(["Score", "Identifier"], ascending=[False, True])
    if params is None or params.get("deduplicate_by_inchikey1", True):
        res = res.drop_duplicates("InChIKey1", keep="first")
    top_n = 50 if params is None else int(params.get("top_n_candidates", 50))
    res = res.head(top_n).copy()

    # 去重 Identifier
    seen = {}
    ids = []

    for x in res["Identifier"]:
        x = str(x)
        if x in seen:
            seen[x] += 1
            ids.append(f"{x}_{seen[x]}")
        else:
            seen[x] = 0
            ids.append(x)

    res["Identifier"] = ids

    res["cluster index"] = int(node_id)
    res["uid"] = res.apply(
        lambda a: f"{int(node_id)}_{a['Identifier']}",
        axis=1
    )

    clean_res = res[[
        "uid",
        "cluster index",
        "Identifier",
        "InChI",
        "InChIKey",
        "InChIKey1",
        "Score",
        "Score_Clean"
    ]].copy()

    print("\n清洗后的候选物检查：")
    display(clean_res.head(10))

    return clean_res


# =========================================================
# 种子节点清洗
# =========================================================

def prepare_seed_table(seeds_df):
    seeds = seeds_df.copy()
    seeds.columns = [str(c).strip() for c in seeds.columns]

    if "#Scan#" in seeds.columns and "cluster index" not in seeds.columns:
        seeds["cluster index"] = seeds["#Scan#"].astype(int)

    if "cluster index" not in seeds.columns:
        raise ValueError("种子文件缺少 cluster index 或 #Scan# 列。")

    if "Identifier" not in seeds.columns:
        seeds["Identifier"] = [f"Seed_{i}" for i in range(len(seeds))]

    seeds["cluster index"] = seeds["cluster index"].astype(int)
    seeds["Identifier"] = seeds["Identifier"].astype(str)

    # InChI 优先；没有的话从 SMILES 转
    if "InChI" not in seeds.columns:
        seeds["InChI"] = ""

    bad_inchi = ~seeds["InChI"].map(is_usable_inchi)

    if bad_inchi.any():
        smiles_col = find_smiles_col(seeds)
        if smiles_col is not None:
            seeds.loc[bad_inchi, "InChI"] = (
                seeds.loc[bad_inchi, smiles_col].apply(smiles_to_inchi)
            )

    bad_inchi = ~seeds["InChI"].map(is_usable_inchi)

    if bad_inchi.any():
        print("🚨 以下种子没有有效 InChI 或 SMILES：")
        _show(seeds.loc[bad_inchi])
        raise ValueError(
            "种子节点必须提供有效 InChI 或 SMILES。"
            "只有 cluster index、Identifier、Score 或 InChIKey1 是不够的。"
        )

    print("\n种子节点检查：")
    _show(seeds[["cluster index", "Identifier", "InChI"]])

    # 给种子加 Seed_ 前缀
    seeds["Identifier"] = seeds["Identifier"].apply(
        lambda x: x if str(x).startswith("Seed_") else f"Seed_{x}"
    )

    seeds["BaseIdentifier"] = seeds["Identifier"].str.replace("^Seed_", "", regex=True)

    seeds["uid"] = seeds.apply(
        lambda a: f"{a['cluster index']}_{a['Identifier']}",
        axis=1
    )

    seeds["Score_Clean"] = 1.0
    seeds["Score"] = 1.0

    return seeds[
        ["uid", "cluster index", "Identifier", "BaseIdentifier",
         "InChI", "Score", "Score_Clean"]
    ].copy()


# =========================================================
# RDKit 结构相似度
# =========================================================

def get_fingerprint_from_inchi(inchi, method="MFP2-bits"):
    """生成与基准评估相同定义的 RDKit 结构指纹。"""
    mol = mol_from_inchi(inchi)
    if mol is None:
        return None

    method = str(method).upper()

    if method == "MACCS":
        return MACCSkeys.GenMACCSKeys(mol)
    if method == "MFP2-BITS":
        return rdMolDescriptors.GetHashedMorganFingerprint(mol, 2)
    if method == "MFP1-BITS":
        return rdMolDescriptors.GetHashedMorganFingerprint(mol, 1)
    if method == "MFP2":
        return rdMolDescriptors.GetMorganFingerprint(mol, 2)
    if method == "MFP1":
        return rdMolDescriptors.GetMorganFingerprint(mol, 1)
    if method == "RDKIT7-LINEAR":
        return Chem.RDKFingerprint(mol, maxPath=7, branchedPaths=False)
    if method.startswith("RDKIT"):
        nums = re.findall(r"\d+", method)
        max_path = int(nums[0]) if len(nums) > 0 else 7
        return Chem.RDKFingerprint(mol, maxPath=max_path)

    raise ValueError(f"不支持的结构指纹方法：{method}")


def chemwalker_sigmoid(x, a=-9, b=0.60):
    """
    ChemWalker/benchmark 中使用的 sigmoid 转换。
    x 通常为 Cosine × Tanimoto。
    """
    return 1.0 / (1.0 + np.exp(a * (x - b)))


def run_random_walk_consistent(G, seed_uids, restart_prob=0.5):
    """
    与正式 benchmark 保持一致的 random_walk 调用。
    不再使用两个节点 direct mode。
    """
    seed_uids = [x for x in seed_uids if x in G.nodes]

    if len(seed_uids) == 0:
        raise ValueError("没有种子节点连入候选物网络，无法进行 random walk。")

    p_t = random_walk(
        G,
        seed_uids,
        restart_prob=restart_prob
    )

    if isinstance(p_t, dict):
        return pd.DataFrame(
            list(p_t.items()),
            columns=["uid", "mNAP_score"]
        )

    nodes = list(G.nodes())

    return pd.DataFrame(
        zip(nodes, p_t),
        columns=["uid", "mNAP_score"]
    )


def build_similarity_edges(
    snet,
    full_tlid,
    method="MFP2-bits",
    edge_mode="sigmoid",
    min_weight=0.0,
    use_metfrag_in_edges=True,
    metfrag_beta=0.5,
    score_eps=0.05,
    sigmoid_a=-9,
    sigmoid_b=0.60
):
    """
    与正式 benchmark 一致的 mNAP 候选物边权：

    edge_mode = "sigmoid":
        w = sigmoid(Cosine × Tanimoto) × (S_i* S_j*)^beta

    其中 S* = max(S, score_eps)，避免极低 MetFrag 分数使边权完全消失。
    """

    full_tlid = full_tlid.copy()

    # 每个 cluster 内部归一化 MetFrag score
    full_tlid["Score"] = pd.to_numeric(
        full_tlid["Score"],
        errors="coerce"
    ).fillna(0)

    full_tlid["metfrag_score_norm"] = full_tlid.groupby("cluster index")["Score"].transform(
        lambda x: x / x.max() if x.max() > 0 else 0
    )

    # 种子节点强制设为 1
    full_tlid.loc[
        full_tlid["Identifier"].astype(str).str.startswith("Seed_"),
        "metfrag_score_norm"
    ] = 1.0

    fps = {}

    for _, row in full_tlid.iterrows():
        uid = row["uid"]
        fps[uid] = get_fingerprint_from_inchi(row["InChI"], method=method)

    edges = []

    for _, e in snet.iterrows():
        c1 = int(e["CLUSTERID1"])
        c2 = int(e["CLUSTERID2"])

        cosine = float(e["Cosine"]) if "Cosine" in e and pd.notna(e["Cosine"]) else 0.0

        df1 = full_tlid[full_tlid["cluster index"] == c1]
        df2 = full_tlid[full_tlid["cluster index"] == c2]

        for _, r1 in df1.iterrows():
            uid1 = r1["uid"]
            fp1 = fps.get(uid1)

            if fp1 is None:
                continue

            s1 = float(r1.get("metfrag_score_norm", 0))
            s1_eff = max(s1, score_eps)

            for _, r2 in df2.iterrows():
                uid2 = r2["uid"]
                fp2 = fps.get(uid2)

                if fp2 is None:
                    continue

                s2 = float(r2.get("metfrag_score_norm", 0))
                s2_eff = max(s2, score_eps)

                tanimoto_sim = DataStructs.TanimotoSimilarity(fp1, fp2)

                if edge_mode == "plain":
                    sim_term = tanimoto_sim

                elif edge_mode == "cosine":
                    sim_term = cosine * tanimoto_sim

                elif edge_mode == "sigmoid":
                    sim_term = chemwalker_sigmoid(
                        cosine * tanimoto_sim,
                        a=sigmoid_a,
                        b=sigmoid_b
                    )

                else:
                    raise ValueError("edge_mode must be 'plain', 'cosine', or 'sigmoid'.")

                if use_metfrag_in_edges:
                    weight = sim_term * (s1_eff ** metfrag_beta) * (s2_eff ** metfrag_beta)
                else:
                    weight = sim_term

                if weight > min_weight and not pd.isna(weight):
                    edges.append({
                        "u": uid1,
                        "v": uid2,
                        "edge_weight": float(weight),
                        "struct_similarity": float(tanimoto_sim),
                        "cosine": float(cosine),
                        "sim_term": float(sim_term),
                        "metfrag_i": float(s1),
                        "metfrag_j": float(s2),
                        "metfrag_i_eff": float(s1_eff),
                        "metfrag_j_eff": float(s2_eff)
                    })

    return pd.DataFrame(edges)


def add_ranks_and_normalize(final_df):
    final_df = final_df.copy()

    final_df["mNAP_score"] = pd.to_numeric(
        final_df["mNAP_score"], errors="coerce"
    ).fillna(0.0)

    group_max = final_df.groupby("cluster index")["mNAP_score"].transform("max")
    final_df["mNAP_reachable"] = group_max > 0
    final_df["mNAP_score_norm"] = np.where(
        group_max > 0,
        final_df["mNAP_score"] / group_max,
        0.0
    )

    final_df["mNAP_rank"] = final_df.groupby("cluster index")["mNAP_score_norm"].rank(
        ascending=False,
        method="dense"
    )
    final_df.loc[~final_df["mNAP_reachable"], "mNAP_rank"] = np.nan

    final_df["MetFrag_rank"] = final_df.groupby("cluster index")["Score"].rank(
        ascending=False,
        method="dense"
    )

    return final_df


# =========================================================
# 主函数
# =========================================================

def mNAP(component_nodes, seeds_df, net_df, spectra_list, db_df, params):
    tlid_all = []

    component_nodes = [int(x) for x in component_nodes]

    seeds_df = seeds_df.copy()
    seeds_df.columns = [str(c).strip() for c in seeds_df.columns]

    if "#Scan#" in seeds_df.columns and "cluster index" not in seeds_df.columns:
        seeds_df["cluster index"] = seeds_df["#Scan#"].astype(int)

    seed_nodes_set = set(seeds_df["cluster index"].astype(int))

    print(f"开始为组件中的 {len(component_nodes)} 个节点获取候选物...")

    # -------------------------
    # 1. 未知节点：跑 MetFrag
    # -------------------------
    for node_id in component_nodes:
        if int(node_id) in seed_nodes_set:
            print(f"  ⏩ 节点 {node_id} 是已知种子节点，跳过 MetFrag 盲搜环节。")
            continue

        try:
            print(f"  -> 正在对未知节点 {node_id} 进行 MetFrag 碎片打分...")

            spec_list = [
                s for s in spectra_list
                if str(s["params"].get("scans")) == str(node_id)
            ]

            if len(spec_list) == 0:
                print(f"⚠️ 找不到节点 {node_id} 对应的谱图，跳过。")
                continue

            spec = spec_list[0]

            res = run_metfrag(
                spec,
                db_df,
                node_id,
                metpath=params["metpath"],
                ppm=params.get("ppm", 10)
            )

            if res is None or res.empty:
                print(f"⚠️ 节点 {node_id} 没有 MetFrag 候选物。")
                continue

            clean_res = clean_metfrag_result(
                res=res,
                node_id=node_id,
                db_df=db_df,
                params=params,
                verbose=True
            )

            if clean_res.empty:
                print(f"🚨 节点 {node_id} 没有任何带有效结构的候选物。")
                continue

            tlid_all.append(clean_res)

        except Exception as e:
            print(f"⚠️ 节点 {node_id} 处理异常: {e}")

    # -------------------------
    # 2. 种子节点
    # -------------------------
    print("\n正在将种子节点结构汇入候选物总表...")
    seeds_clean = prepare_seed_table(seeds_df)
    tlid_all.append(seeds_clean)

    if not tlid_all:
        print("🚨 没有提取到任何有效候选物或种子！")
        return pd.DataFrame()

    full_tlid = pd.concat(tlid_all, sort=False).reset_index(drop=True)

    print("\n候选物总表检查：")
    _show(full_tlid[["uid", "cluster index", "Identifier", "InChI", "Score"]].head(30))

    # 强制检查：不允许 InChI=2 这类错误进入后续计算
    bad_all = ~full_tlid["InChI"].map(is_usable_inchi)

    if bad_all.any():
        print("\n🚨 full_tlid 中仍存在无效 InChI，不能继续。")
        _show(full_tlid.loc[bad_all, ["uid", "Identifier", "InChI"]].head(30))
        raise ValueError("full_tlid 中存在无效 InChI，请检查 MetFrag 输出或 db 结构列。")

    # -------------------------
    # 3. 子网络
    # -------------------------
    net_df = net_df.copy()
    net_df["CLUSTERID1"] = net_df["CLUSTERID1"].astype(int)
    net_df["CLUSTERID2"] = net_df["CLUSTERID2"].astype(int)

    snet = net_df[
        net_df["CLUSTERID1"].isin(component_nodes)
        & net_df["CLUSTERID2"].isin(component_nodes)
    ].copy()

    if snet.empty:
        display(net_df[["CLUSTERID1", "CLUSTERID2", "Cosine", "DeltaMZ"]].head())
        raise ValueError(
            "子网络为空。说明 component_nodes 和 net_df 中的 CLUSTERID1/CLUSTERID2 没有匹配上。"
        )

    print("\n子网络边：")
    _show(snet)

    # -------------------------
    # 4. 自己计算候选物结构相似度边
    # -------------------------
    method = params.get("structure_method", "MFP2-bits")
    edge_mode = params.get("edge_mode", "sigmoid")
    
    print(
        f"\n正在计算 mNAP 候选物边权："
        f"method={method}, edge_mode={edge_mode}, "
        f"beta={params.get('metfrag_beta', 0.5)}, "
        f"score_eps={params.get('score_eps', 0.05)}"
    )
    
    scandpair = build_similarity_edges(
        snet=snet,
        full_tlid=full_tlid,
        method=method,
        edge_mode=edge_mode,
        min_weight=0.0,
        use_metfrag_in_edges=params.get("use_metfrag_in_edges", True),
        metfrag_beta=params.get("metfrag_beta", 0.5),
        score_eps=params.get("score_eps", 0.05),
        sigmoid_a=params.get("sigmoid_a", -9),
        sigmoid_b=params.get("sigmoid_b", 0.60)
    )
    if scandpair.empty:
        raise ValueError(
            "没有生成任何候选物相似度边。可能原因："
            "1）候选物 InChI 无效；"
            "2）结构相似度全为 0；"
            "3）use_metfrag_in_edges=True 时某些节点 MetFrag score 全为 0；"
            "4）种子节点没有和未知节点连上。"
        )

    # -------------------------
    # 5. 先看种子到候选物的相似度
    # -------------------------
    seed_uids = full_tlid.loc[
        full_tlid["Identifier"].astype(str).str.startswith("Seed_"),
        "uid"
    ].tolist()

    debug_edges = scandpair[
        scandpair["u"].isin(seed_uids) | scandpair["v"].isin(seed_uids)
    ].copy()

    if not debug_edges.empty:
        debug_edges["candidate_uid"] = debug_edges.apply(
            lambda x: x["v"] if x["u"] in seed_uids else x["u"],
            axis=1
        )

        print("\n种子到候选物的结构相似度排序：")
        _show(
            debug_edges[
                ["candidate_uid", "edge_weight", "struct_similarity", "cosine"]
            ].sort_values("edge_weight", ascending=False).head(30)
        )
    else:
        print("seed_uids:")
        print(seed_uids)
    
        print("scandpair 前 20 行：")
        display(scandpair.head(20))
    
        raise ValueError(
            "没有找到种子到候选物的边。说明种子虽然在 full_tlid 中，但没有进入候选物边网络。"
        )

    # -------------------------
    # 6. 统一使用 random walk，与正式 benchmark 保持一致
    # -------------------------
    print("\n✅ 开始基于 mNAP 边权进行种子信息扩散。")
    
    G = nx.Graph()
    G.add_weighted_edges_from(
        scandpair[["u", "v", "edge_weight"]]
        .itertuples(index=False, name=None)
    )
    
    source = [
        n for n in G.nodes()
        if "_Seed_" in str(n)
    ]
    
    if not source:
        print("G.nodes 前 20 个：")
        print(list(G.nodes())[:20])
    
        raise ValueError(
            "种子未能连入 random walk 网络。请检查 seed uid 是否包含 '_Seed_'，以及种子是否有非零边权。"
        )
    
    results = run_random_walk_consistent(
        G,
        source,
        restart_prob=params.get("restart_prob", 0.5)
    )
    
    final_df = pd.merge(
        full_tlid,
        results,
        on="uid",
        how="left"
    ).fillna(0)
    
    final_df = add_ranks_and_normalize(final_df)
    
    return final_df.sort_values(
        ["cluster index", "mNAP_score"],
        ascending=[True, False]
    )
    
```

```python
from rdkit import Chem
from rdkit.Chem import Draw
from rdkit.Chem.Draw import IPythonConsole
import pandas as pd


def draw_comparison_panels(result_df, target_scan, top_n=10):
    """
    绘制：
    1. 原始 MetFrag Top N
    2. 网络传播 Top N

    注意：
    如果 use_metfrag_in_edges=True，
    那么 mNAP_score 已经包含 MetFrag score 对传播路径的影响。
    """

    df_node = result_df[result_df["cluster index"] == int(target_scan)].copy()

    df_node = df_node[
        ~df_node["Identifier"].astype(str).str.startswith("Seed_")
    ].copy()

    if df_node.empty:
        print(f"❌ 未找到节点 {target_scan} 的候选物。")
        return

    def generate_grid(df, sort_col, title_prefix):
        if sort_col not in df.columns:
            print(f"❌ 缺少排序列：{sort_col}")
            return None

        top_df = df.sort_values(by=sort_col, ascending=False).head(top_n)

        mols = []
        legends = []

        for _, row in top_df.iterrows():
            inchi = str(row.get("InChI", "")).strip()

            if not inchi.startswith("InChI="):
                continue

            mol = Chem.MolFromInchi(inchi)

            if mol is None:
                continue

            mols.append(mol)

            identifier = row.get("Identifier", "")
            metfrag_score = row.get("Score", 0)
            mnap_score = row.get("mNAP_score", 0)

            legends.append(
                f"ID: {identifier}\n"
                f"MetFrag: {metfrag_score:.6f}\n"
                f"mNAP: {mnap_score:.6f}"
            )

        print(f"\n--- 节点 {target_scan}：{title_prefix} Top {top_n} ---")

        if len(mols) == 0:
            print(f"❌ 前 {top_n} 个候选物中没有可绘制的有效 InChI。")
            return None

        return Draw.MolsToGridImage(
            mols,
            molsPerRow=5,
            subImgSize=(260, 250),
            legends=legends
        )

    img_metfrag = generate_grid(df_node, "Score", "原始 MetFrag")
    if img_metfrag is not None:
        display(img_metfrag)

    img_network = generate_grid(df_node, "mNAP_score", "mNAP 重排后")
    if img_network is not None:
        display(img_network)


target_node = 88874
draw_comparison_panels(result_tab, target_node, top_n=5)
```
